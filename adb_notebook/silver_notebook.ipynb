```python
from pyspark.sql.functions import *
```

### Data Reading

```python
bronze_df = spark.read.format("parquet")\
            .option("inferschema", "true")\
            .load("/Volumes/adbkomal/bronze/raw_data/CarSalesRawData/")
display(bronze_df)
```

### Data Transformation

```python
bronze_df = bronze_df.withColumn("Model_Category", split(col("Model_ID"), "-")[0])
display(bronze_df)
```

```python
# bronze_df.printSchema()
bronze_df.withColumn("Units_Sold", col("Units_Sold").cast("String")).printSchema()
```

```python
bronze_df = bronze_df.withColumn("revenue_per_unit", col("Revenue") / col("Units_Sold"))
display(bronze_df)
```

**aggregation functions**
- how many units got sold of each branch each year
- show results in year acending and total units sold in descending order

```python
display(bronze_df.groupBy("Year","BranchName").agg(sum(col("Units_Sold")).alias("Total_Sold")).sort(col("Year"), col("Total_sold"), ascending=[True, False]))
```

```python
bronze_df.write.format("parquet")\
    .mode("overwrite")\
    .save("/Volumes/adbkomal/silver/silver_data/CarSalesData")
```
